```python
import re
import pandas as pd

def normalize_text(x):
    if pd.isna(x):
        return ""

    x = str(x).lower()
    x = x.replace("&", " and ")
    x = re.sub(r"[^a-z0-9\s]", " ", x)
    x = re.sub(r"\s+", " ", x).strip()

    return x
```

```python
import pandas as pd
from collections import defaultdict

s3_block_index = defaultdict(list)

for chunk in pd.read_csv(
    S3_PATH,
    sep="\t",
    chunksize=200_000,
    usecols=["entity_id", "business_name", "business_address", "country"]
):
    chunk["name_norm"] = chunk["business_name"].apply(normalize_text)
    chunk["address_norm"] = chunk["business_address"].apply(normalize_text)

    chunk["name_prefix4"] = chunk["name_norm"].str[:4]
    chunk["address_prefix4"] = chunk["address_norm"].str[:4]

    for row in chunk.itertuples(index=False):
        key = (
            str(row.country)
            + "|"
            + row.name_prefix4
            + "|"
            + row.address_prefix4
        )

        s3_block_index[key].append(row.entity_id)

print("S3 block index keys:", len(s3_block_index))
```

```python
import numpy as np

block_sizes = np.array([
    len(v) for v in s3_block_index.values()
])

print("Total block keys:", len(block_sizes))
print("Total S3 rows in index:", block_sizes.sum())
print("Average rows per block:", block_sizes.mean())
print("Median rows per block:", np.median(block_sizes))
print("Max rows in a block:", block_sizes.max())

print("\nBlock size distribution:")
print("Size = 1 :", np.sum(block_sizes == 1))
print("Size 2-5:", np.sum((block_sizes >= 2) & (block_sizes <= 5)))
print("Size 6-10:", np.sum((block_sizes >= 6) & (block_sizes <= 10)))
print("Size >10:", np.sum(block_sizes > 10))
```

```python
gt_raw = pd.read_csv(
    GT_PATH,
    sep="\t",
    usecols=["source1_entity_id", "matched_entity_ids"]
)

gt_s3_pairs = []

for row in gt_raw.itertuples(index=False):
    matched_ids = str(row.matched_entity_ids).split(",")

    for entity_id in matched_ids:
        entity_id = entity_id.strip()

        if entity_id.startswith("S3-"):
            gt_s3_pairs.append(
                (row.source1_entity_id, entity_id)
            )

gt_s3 = pd.DataFrame(
    gt_s3_pairs,
    columns=["source1_entity_id", "source3_entity_id"]
)

print("Total S1→S3 ground-truth matches:", len(gt_s3))
display(gt_s3.head(10))
```

```python
from collections import defaultdict

s3_exact_name_country_index = defaultdict(list)

for chunk in pd.read_csv(
    S3_PATH,
    sep="\t",
    chunksize=200_000,
    usecols=["entity_id", "business_name", "country"]
):
    chunk["name_norm"] = chunk["business_name"].apply(normalize_text)

    for row in chunk.itertuples(index=False):
        if row.name_norm:
            key = str(row.country) + "|" + row.name_norm
            s3_exact_name_country_index[key].append(row.entity_id)

print("S3 country + exact-name keys:", len(s3_exact_name_country_index))
```

```python
# Build S1 lookup: entity_id -> country + normalized business name

s1_lookup = {}

for chunk in pd.read_csv(
    S1_PATH,
    sep="\t",
    chunksize=200_000,
    usecols=["entity_id", "business_name", "country"]
):
    chunk["name_norm"] = chunk["business_name"].apply(normalize_text)

    for row in chunk.itertuples(index=False):
        s1_lookup[row.entity_id] = (
            str(row.country),
            row.name_norm
        )

print("S1 lookup entries:", len(s1_lookup))


# Check exact-name blocking recall

covered = 0

for row in gt_s3.itertuples(index=False):
    country, name_norm = s1_lookup[row.source1_entity_id]

    key = country + "|" + name_norm

    if row.source3_entity_id in s3_exact_name_country_index.get(key, []):
        covered += 1

total = len(gt_s3)
recall = covered / total * 100

print("Total S1→S3 ground-truth pairs:", total)
print("Covered by country + exact-name:", covered)
print("Missed:", total - covered)
print(f"Recall: {recall:.2f}%")
```

```python
# Rebuild S1 lookup with name + address + country

s1_lookup = {}

for chunk in pd.read_csv(
    S1_PATH,
    sep="\t",
    chunksize=200_000,
    usecols=[
        "entity_id",
        "business_name",
        "business_address",
        "country"
    ]
):
    chunk["name_norm"] = chunk["business_name"].apply(normalize_text)
    chunk["address_norm"] = chunk["business_address"].apply(normalize_text)

    for row in chunk.itertuples(index=False):
        s1_lookup[row.entity_id] = (
            str(row.country),
            row.name_norm,
            row.address_norm
        )
S1 lookup entries: 2206821$0
print("S1 lookup entries:", len(s1_lookup))
```

```python
covered_union = 0

for row in gt_s3.itertuples(index=False):
    country, name_norm, address_norm = s1_lookup[row.source1_entity_id]

    # Rule 1: country + first 4 chars of name + first 4 chars of address
    rule1_key = (
        country
        + "|"
        + name_norm[:4]
        + "|"
        + address_norm[:4]
    )

    rule1_match = (
        row.source3_entity_id
        in s3_block_index.get(rule1_key, [])
    )

    # Rule 2: country + exact normalized name
    rule2_key = country + "|" + name_norm

    rule2_match = (
        row.source3_entity_id
        in s3_exact_name_country_index.get(rule2_key, [])
    )

    if rule1_match or rule2_match:
        covered_union += 1

total = len(gt_s3)
missed = total - covered_union
recall = covered_union / total * 100

print("Total S1→S3 ground-truth pairs:", total)
print("Covered by Rule 1 + Rule 2:", covered_union)
print("Missed:", missed)
print(f"Union Recall: {recall:.2f}%")
```

```python
from collections import Counter

s3_token_counts = Counter()

for chunk in pd.read_csv(
    S3_PATH,
    sep="\t",
    chunksize=200_000,
    usecols=["business_name"]
):
    chunk["name_norm"] = chunk["business_name"].apply(normalize_text)

    for name in chunk["name_norm"]:
        tokens = set(name.split())

        for token in tokens:
            if token:
                s3_token_counts[token] += 1

print("Unique S3 name tokens:", len(s3_token_counts))

print("\nToken frequency distribution:")
print("<= 10:", sum(v <= 10 for v in s3_token_counts.values()))
print("<= 100:", sum(v <= 100 for v in s3_token_counts.values()))
print("<= 500:", sum(v <= 500 for v in s3_token_counts.values()))
print("<= 1000:", sum(v <= 1000 for v in s3_token_counts.values()))
print("> 1000:", sum(v > 1000 for v in s3_token_counts.values()))
```

```python
from collections import defaultdict

s3_token_index = defaultdict(list)

for chunk in pd.read_csv(
    S3_PATH,
    sep="\t",
    chunksize=200_000,
    usecols=["entity_id", "business_name", "country"]
):
    chunk["name_norm"] = chunk["business_name"].apply(normalize_text)

    for row in chunk.itertuples(index=False):
        tokens = set(row.name_norm.split())

        for token in tokens:
            if token and s3_token_counts[token] <= 1000:
                key = str(row.country) + "|" + token
                s3_token_index[key].append(row.entity_id)

print("S3 country + token keys:", len(s3_token_index))
```

```python
covered_r3 = 0

for row in gt_s3.itertuples(index=False):
    country, name_norm, address_norm = s1_lookup[row.source1_entity_id]

    tokens = set(name_norm.split())

    matched = False

    for token in tokens:
        if token and s3_token_counts[token] <= 1000:
            key = country + "|" + token

            if row.source3_entity_id in s3_token_index.get(key, []):
                matched = True
                break

    if matched:
        covered_r3 += 1

total = len(gt_s3)
missed_r3 = total - covered_r3
recall_r3 = covered_r3 / total * 100

print("Total S1→S3 ground-truth pairs:", total)
print("Covered by Rule 3:", covered_r3)
print("Missed:", missed_r3)
print(f"Rule 3 Recall: {recall_r3:.2f}%")
```

```python
covered_all = 0

for row in gt_s3.itertuples(index=False):
    country, name_norm, address_norm = s1_lookup[row.source1_entity_id]

    # Rule 1
    rule1_key = country + "|" + name_norm[:4] + "|" + address_norm[:4]
    rule1_match = row.source3_entity_id in s3_block_index.get(rule1_key, [])

    # Rule 2
    rule2_key = country + "|" + name_norm
    rule2_match = row.source3_entity_id in s3_exact_name_country_index.get(rule2_key, [])

    # Rule 3
    rule3_match = False
    for token in set(name_norm.split()):
        if token and s3_token_counts[token] <= 1000:
            rule3_key = country + "|" + token

            if row.source3_entity_id in s3_token_index.get(rule3_key, []):
                rule3_match = True
                break

    if rule1_match or rule2_match or rule3_match:
        covered_all += 1

total = len(gt_s3)
missed = total - covered_all
recall = covered_all / total * 100

print("Total S1→S3 ground-truth pairs:", total)
print("Covered by R1 + R2 + R3:", covered_all)
print("Missed:", missed)
print(f"Final blocking union recall: {recall:.2f}%")
```

```python
total_candidates = 0
s1_count = 0

for row in s1_lookup.values():
    country, name_norm, address_norm = row

    candidate_ids = set()

    # Rule 1
    rule1_key = country + "|" + name_norm[:4] + "|" + address_norm[:4]
    candidate_ids.update(s3_block_index.get(rule1_key, []))

    # Rule 2
    rule2_key = country + "|" + name_norm
    candidate_ids.update(s3_exact_name_country_index.get(rule2_key, []))

    # Rule 3
    for token in set(name_norm.split()):
        if token and s3_token_counts[token] <= 1000:
            rule3_key = country + "|" + token
            candidate_ids.update(s3_token_index.get(rule3_key, []))

    total_candidates += len(candidate_ids)
    s1_count += 1

print("S1 records:", s1_count)
print("Estimated candidate pairs:", total_candidates)
print("Average candidates per S1:", total_candidates / s1_count)
```

```python
import csv
import os
import time
import pandas as pd

FINAL_BLOCK_FILE = (
    "/content/drive/MyDrive/Amazon_ML_Challenge_2026/"
    "s3_blocking/final_s3_blocking.tsv"
)

# Safety: overwrite existing test/partial file
if os.path.exists(FINAL_BLOCK_FILE):
    os.remove(FINAL_BLOCK_FILE)

# Read S1 in chunks
S1_CHUNK_SIZE = 10_000

start_time = time.time()
total_pairs = 0
total_s1 = 0

with open(
    FINAL_BLOCK_FILE,
    "w",
    newline="",
    encoding="utf-8",
    buffering=1024 * 1024
) as f:

    writer = csv.writer(f, delimiter="\t")

    # Header
    writer.writerow([
        "source1_entity_id",
        "source3_entity_id"
    ])

    for chunk_no, chunk in enumerate(
        pd.read_csv(
            S1_PATH,
            sep="\t",
            chunksize=S1_CHUNK_SIZE,
            usecols=[
                "entity_id",
                "business_name",
                "business_address",
                "country"
            ]
        ),
        start=1
    ):

        chunk["name_norm"] = (
            chunk["business_name"]
            .apply(normalize_text)
        )

        chunk["address_norm"] = (
            chunk["business_address"]
            .apply(normalize_text)
        )

        chunk_pairs = 0

        for row in chunk.itertuples(index=False):

            s1_id = row.entity_id
            country = str(row.country)
            name_norm = row.name_norm
            address_norm = row.address_norm

            candidate_ids = set()

            # ------------------------------------------------
            # Rule 1
            # country + name prefix4 + address prefix4
            # ------------------------------------------------
            rule1_key = (
                country
                + "|"
                + name_norm[:4]
                + "|"
                + address_norm[:4]
            )

            candidate_ids.update(
                s3_block_index.get(rule1_key, [])
            )

            # ------------------------------------------------
            # Rule 2
            # country + exact normalized business name
            # ------------------------------------------------
            if name_norm:
                rule2_key = country + "|" + name_norm

                candidate_ids.update(
                    s3_exact_name_country_index.get(
                        rule2_key,
                        []
                    )
                )

            # ------------------------------------------------
            # Rule 3
            # country + name token
            # frequency <= 1000
            # ------------------------------------------------
            for token in set(name_norm.split()):

                if (
                    token
                    and s3_token_counts[token] <= 1000
                ):
                    rule3_key = country + "|" + token

                    candidate_ids.update(
                        s3_token_index.get(
                            rule3_key,
                            []
                        )
                    )

            # Write deduplicated S1-S3 candidates
            for s3_id in candidate_ids:
                writer.writerow([
                    s1_id,
                    s3_id
                ])

            chunk_pairs += len(candidate_ids)

        total_pairs += chunk_pairs
        total_s1 += len(chunk)

        elapsed = time.time() - start_time

        print(
            f"Chunk {chunk_no:4d} | "
            f"S1 processed: {total_s1:,} | "
            f"Pairs written: {total_pairs:,} | "
            f"Chunk pairs: {chunk_pairs:,} | "
            f"Elapsed: {elapsed/60:.1f} min"
        )

print("\n========================================")
print("FINAL S3 BLOCKING COMPLETED")
print("========================================")
print("S1 records processed:", f"{total_s1:,}")
print("Final candidate pairs:", f"{total_pairs:,}")
print("Output file:")
print(FINAL_BLOCK_FILE)

file_size_gb = os.path.getsize(FINAL_BLOCK_FILE) / (1024**3)

print(f"File size: {file_size_gb:.2f} GB")
print(
    "Average candidates per S1:",
    total_pairs / total_s1
)
```
