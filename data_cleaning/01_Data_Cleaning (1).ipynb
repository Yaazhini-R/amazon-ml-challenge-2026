```python
import unicodedata
import re

def normalize_unicode_text(text):
    if pd.isna(text):
        return text

    text = str(text).lower()

    # Normalize Unicode representation
    text = unicodedata.normalize("NFKC", text)

    # Replace punctuation/symbols with spaces,
    # while preserving letters and numbers from all languages
    result = []

    for char in text:
        category = unicodedata.category(char)

        if category[0] in ("L", "N"):
            # Letter or number
            result.append(char)
        else:
            # Punctuation, symbols, marks, etc.
            result.append(" ")

    text = "".join(result)

    # Remove extra whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
def normalize_unicode_text(text):
    if pd.isna(text):
        return text

    text = str(text).lower()

    # Normalize Unicode representation
    text = unicodedata.normalize("NFKC", text)

    result = []

    for char in text:
        category = unicodedata.category(char)

        # Keep letters, numbers and combining marks
        if category[0] in ("L", "N", "M"):
            result.append(char)
        else:
            # Replace punctuation/symbols with space
            result.append(" ")

    text = "".join(result)

    # Remove repeated whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
import unicodedata
import re

def normalize_unicode_text(text):
    if pd.isna(text):
        return text

    text = str(text).lower()

    # Normalize Unicode representation
    text = unicodedata.normalize("NFKC", text)

    result = []

    for char in text:
        category = unicodedata.category(char)

        # Keep letters, numbers and combining marks
        if category[0] in ("L", "N", "M"):
            result.append(char)
        else:
            # Replace punctuation/symbols with space
            result.append(" ")

    text = "".join(result)

    # Remove repeated whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
import unicodedata
import re

def normalize_unicode_text(text):
    if pd.isna(text):
        return text

    text = str(text).lower()

    # Standardize Unicode representation
    text = unicodedata.normalize("NFKC", text)

    result = []

    for char in text:
        category = unicodedata.category(char)

        # Keep letters, numbers and combining marks
        if category[0] in ("L", "N", "M"):
            result.append(char)
        else:
            result.append(" ")

    text = "".join(result)

    # Standardize whitespace
    text = re.sub(r"\s+", " ", text).strip()

    return text
```

```python
import pandas as pd
import re
import unicodedata
import os
import gc
import shutil

# ============================================================
# CLEANING FUNCTIONS
# ============================================================

def clean_control_chars(text):
    if pd.isna(text):
        return text

    text = str(text)
    text = re.sub(r"[\x00-\x1F\x7F]", " ", text)
    text = re.sub(r"\s+", " ", text)

    return text.strip()


def repair_mojibake(text):
    if pd.isna(text):
        return text

    text = str(text)

    replacements = {
        "Â\x80\x93": "–",
        "Â\x80\x99": "’",
        "Â\x80\x98": "‘",
        "Â\x80\x90": "‐",
        "â\x80\x93": "–",
        "â\x80\x99": "’",
        "â\x80\x98": "‘",
        "â\x80\x90": "‐"
    }

    for bad, good in replacements.items():
        text = text.replace(bad, good)

    return text


def normalize_unicode_text(text):
    if pd.isna(text):
        return text

    text = str(text).lower()
    text = unicodedata.normalize("NFKC", text)

    result = []

    for char in text:
        category = unicodedata.category(char)

        if category[0] in ("L", "N", "M"):
            result.append(char)
        else:
            result.append(" ")

    text = "".join(result)
    text = re.sub(r"\s+", " ", text)

    return text.strip()


def clean_chunk(df):

    df["business_name_clean"] = (
        df["business_name"]
        .apply(repair_mojibake)
        .apply(clean_control_chars)
    )

    df["business_address_clean"] = (
        df["business_address"]
        .apply(repair_mojibake)
        .apply(clean_control_chars)
    )

    df["business_name_normalized"] = (
        df["business_name_clean"]
        .apply(normalize_unicode_text)
    )

    df["business_address_normalized"] = (
        df["business_address_clean"]
        .apply(normalize_unicode_text)
    )

    return df


# ============================================================
# PATHS
# ============================================================

base_dir = "/content/drive/MyDrive/Business_model_amazon"

input_file = (
    base_dir +
    "/extracted_data/student_resource/dataset/train/"
    "train_source3.tsv"
)

drive_output = (
    base_dir +
    "/cleaned_data/train_source3_cleaned.tsv"
)

# Write temporary file locally
local_temp = "/content/train_source3_cleaned_TEMP.tsv"

EXPECTED_ROWS = 5_285_603
CHUNK_SIZE = 250_000


# ============================================================
# CHECK ORIGINAL
# ============================================================

if not os.path.exists(input_file):
    raise FileNotFoundError(
        f"Original Train Source 3 not found:\n{input_file}"
    )

print("✅ Original Train Source 3 found")
print(f"Expected rows: {EXPECTED_ROWS:,}")
print()


# ============================================================
# REMOVE OLD LOCAL TEMP
# ============================================================

if os.path.exists(local_temp):
    os.remove(local_temp)


# ============================================================
# PROCESS
# ============================================================

first_chunk = True
total_rows = 0

for chunk_no, chunk in enumerate(
    pd.read_csv(
        input_file,
        sep="\t",
        chunksize=CHUNK_SIZE
    ),
    start=1
):

    chunk = clean_chunk(chunk)

    chunk.to_csv(
        local_temp,
        sep="\t",
        index=False,
        mode="w" if first_chunk else "a",
        header=first_chunk
    )

    total_rows += len(chunk)

    print(
        f"Chunk {chunk_no}: "
        f"{len(chunk):,} rows | "
        f"Total: {total_rows:,}"
    )

    first_chunk = False

    del chunk
    gc.collect()


# ============================================================
# VERIFY LOCAL FILE
# ============================================================

print("\n")
print("=" * 70)
print("LOCAL PROCESSING COMPLETE")
print("=" * 70)

print(f"Rows processed : {total_rows:,}")
print(f"Expected rows  : {EXPECTED_ROWS:,}")

if total_rows != EXPECTED_ROWS:

    print("\n❌ ROW COUNT ERROR")
    print("Google Drive file will NOT be replaced.")

    if os.path.exists(local_temp):
        os.remove(local_temp)

    raise ValueError(
        f"Expected {EXPECTED_ROWS:,} rows "
        f"but processed {total_rows:,}"
    )

print("✅ Row count verification PASSED")


# ============================================================
# LOCAL FILE CHECK
# ============================================================

if not os.path.exists(local_temp):
    raise FileNotFoundError(
        "Local temporary file was not created."
    )

local_size = os.path.getsize(local_temp)

print(
    f"Local temporary file size: "
    f"{local_size / (1024**3):.2f} GB"
)


# ============================================================
# COLUMN CHECK
# ============================================================

check = pd.read_csv(
    local_temp,
    sep="\t",
    nrows=2
)

expected_columns = [
    "entity_id",
    "business_name",
    "business_address",
    "country",
    "business_name_clean",
    "business_address_clean",
    "business_name_normalized",
    "business_address_normalized"
]

if list(check.columns) != expected_columns:

    raise ValueError(
        "❌ Column verification failed.\n"
        f"Got: {list(check.columns)}"
    )

print("✅ Column verification PASSED")


# ============================================================
# COPY TO DRIVE
# ============================================================

print("\nCopying completed file to Google Drive...")
print("Please wait...")

if os.path.exists(drive_output):
    os.remove(drive_output)

shutil.copy2(
    local_temp,
    drive_output
)

print("✅ Copy completed")


# ============================================================
# VERIFY DRIVE FILE
# ============================================================

if not os.path.exists(drive_output):
    raise FileNotFoundError(
        "❌ Final Drive file was not created."
    )

drive_size = os.path.getsize(drive_output)

print(
    f"Drive file size: "
    f"{drive_size / (1024**3):.2f} GB"
)

print("\nVerifying final Drive row count...")

drive_rows = sum(
    1 for _ in open(
        drive_output,
        encoding="utf-8"
    )
) - 1

print(f"Drive rows    : {drive_rows:,}")
print(f"Expected rows : {EXPECTED_ROWS:,}")


if drive_rows != EXPECTED_ROWS:

    raise ValueError(
        f"❌ FINAL DRIVE ROW COUNT ERROR!\n"
        f"Expected: {EXPECTED_ROWS:,}\n"
        f"Got:      {drive_rows:,}"
    )


# ============================================================
# CLEAN LOCAL TEMP
# ============================================================

os.remove(local_temp)


# ============================================================
# SUCCESS
# ============================================================

print("\n")
print("=" * 70)
print("🎉 TRAIN SOURCE 3 REBUILD SUCCESSFUL")
print("=" * 70)

print(f"Rows: {drive_rows:,}")
print("File: train_source3_cleaned.tsv")
print("Location: Google Drive / cleaned_data")

print("\n⛔ DO NOT START EDA YET")
print("Next step: final verification of all 6 files.")
print("=" * 70)
```
